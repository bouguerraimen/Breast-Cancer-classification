```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
breast_cancer_dataset=pd.read_csv("data.csv")
```

```python
print(breast_cancer_dataset)
```

```python
breast_cancer_dataset.head()
```

```python
breast_cancer_dataset.shape
```

```python
breast_cancer_dataset = breast_cancer_dataset.drop(columns=["Unnamed: 32", "id"])
```

```python
breast_cancer_dataset.shape
```

```python
breast_cancer_dataset["diagnosis"] = breast_cancer_dataset["diagnosis"].map({"M": 0, "B": 1})
```

```python
breast_cancer_dataset.info()
```

```python
breast_cancer_dataset.isnull().sum()
```

```python
breast_cancer_dataset.describe()
```

```python
breast_cancer_dataset['diagnosis'].value_counts()
```

```python
breast_cancer_dataset.groupby('diagnosis').mean()
```

```python

X = breast_cancer_dataset.drop("diagnosis", axis=1)
Y = breast_cancer_dataset["diagnosis"]
```

```python
print(X)
```

```python
print(Y)
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=2)
```

```python
print(X.shape,X_train.shape,X_test.shape)
```

```python
model=LogisticRegression()
```

```python
model.fit(X_train,Y_train)
```

```python
X_train_prediction=model.predict(X_train)
```

```python
training_data_accuracy=accuracy_score(Y_train,X_train_prediction)
```

```python
print('Accuracy on training data=',training_data_accuracy)
```

```python
X_test_prediction=model.predict(X_test)
test_data_accuracy=accuracy_score(Y_test,X_test_prediction)
```

```python
print('Accuracy on test data=',test_data_accuracy)
```
